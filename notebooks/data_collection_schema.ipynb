1. The LLM is prompted to generate a natural, real-life scenario. The generation process includes specifying:  
   * One of five emotions experienced by the main character  
   * The character’s gender and age group  
   * The type of social interaction that triggered the corresponding emotion in the scenario  

   These parameters are set to increase the diversity of generated scenarios. Without them, variety would be significantly reduced. For example, if gender and age are not specified, most scenarios may default to something like a *19-year-old college student named Sarah*, even under fairly neutral conditions.  

   **Example Scenario:** A middle-aged man feels anger toward his colleague, who consistently fails to meet deadlines, disrupting the team's plans.  

2. Using the same prompt, the LLM generates the scenario from the perspective of the *main character's conversational partner*. However, this person does **not** have full knowledge of the situation—they are only given partial information and act as an external observer.  

3. These scenarios are validated by the LLM to ensure consistency between perspectives and overall coherence.  

4. Two LLMs are then prompted to engage in a dialogue with each other, based on the given scenarios. There are two versions of the prompt for the main character’s conversational partner:  
   * **Empathetic** — explicitly providing the emotional state of the protagonist.  
   * **Non-empathetic** — instructing the LLM to have a low emotional intelligence response.  

### Summary  

For empathetic dialogues, the conversational partner has full awareness of:  
* The protagonist's emotion  
* The context of the dialogue (i.e., what caused the protagonist to feel that emotion)  

This suggests that a final filtering step (either via human annotation or another LLM pass) may be enough to ensure high dataset quality. However, even without additional filtering, the dataset appears visually acceptable.  

Potential filtering steps include:  
1. Verifying whether the protagonist's speech genuinely reflects the emotion specified in the scenario.  
2. Ensuring that the conversational partner behaves in an appropriately **empathetic** or **non-empathetic** manner.
